```python
# dfs.columns
d={'Student_ID':'student_ID','Gender':'gender','EthnicGroup':'ethnicity', 'ParentEduc':'parent_education', 'LunchType':'lunch_type',
       'TestPrep':'test_prep', 'ParentMaritalStatus':'parent_marital_status', 'PracticeSport':'practice_sport', 'IsFirstChild':'is_first_child',
       'NrSiblings':'num_siblings', 'TransportMeans':'transport', 'WklyStudyHours':'wkly_study_hours', 'MathScore':'math_score',
       'ReadingScore':'reading_score', 'WritingScore':'writing_score'}
dfs=dfs.rename(columns=d)
dfs
```

```python
dfs=dfs.drop(['student_ID','transport'],axis=1)
dfs
```

```python
dfs['gender']=dfs['gender'].map({'female':'Female','male':'Male'})
dfs['ethnicity'].unique()
dfs['ethnicity']=dfs['ethnicity'].map({'group C':'Group C', 'group B':'Group B', 'group A':'Group A', 'group D':'Group D', 'group E':'Group E'})
dfs['parent_education'].unique()
dfs['parent_education']=dfs['parent_education'].map({"bachelor's degree":"Bachelor's", 'some college':'Some College', "master's degree":"Master's",
       "associate's degree":"Associate", 'high school':'High School', 'some high school':'Some High School'})
dfs['lunch_type']=dfs['lunch_type'].str.title()
dfs['test_prep']=dfs['test_prep'].str.title()
dfs['parent_marital_status']=dfs['parent_marital_status'].str.title()
dfs['practice_sport']=dfs['practice_sport'].str.title()
dfs['is_first_child']=dfs['is_first_child'].str.title()
dfs
```

```python
cat_cols=['gender', 'ethnicity', 'parent_education', 'lunch_type', 'test_prep',
       'parent_marital_status', 'practice_sport', 'is_first_child']
for col in cat_cols:
  dfs[col]=dfs[col].astype('category')
# dfs.columns
```

```python
dfs['total_score']=(dfs['math_score']+dfs['reading_score']+dfs['writing_score'])
dfs['avarage_score']=dfs['total_score']/3
```

```python
overall_avg = dfs['avarage_score'].mean()
print("Overall Average Score:", overall_avg)
```
